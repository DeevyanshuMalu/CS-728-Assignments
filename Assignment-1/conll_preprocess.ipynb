```python
import json
```

```python
data = []
with open("conll_data/train.jsonl", "r") as f:
	for line in f:
		data.append(json.loads(line))
```

```python
data[0]
```

```python
vocab = {token for line in data for token in line["tokens"]}
len(vocab)
```

```python
vocab_dict = {v:i for i, v in enumerate(vocab)}
len(vocab_dict)
```

```python
with open("conll_data/vocab.json", "w") as f:
	json.dump(vocab_dict, f, indent=4)
```
